Take a dictionary of the product
productname : quantity
1 . allow user to update the quantity of the particular product
2 . display all the product that are less than the given threshold
3 . deletes  a product from the dictionary based on the user input

```python
n=int(input("Enter the value size of the inventory : "))
products={}
for i in range(n):
  name=input("Enter the name of the product  : ")
  quantity=int(input("Enter the quantity of the products : "))
  products[name]=quantity
print(products)
update_product=input("Enter the product to change the quantitiy : ")
update_quantity=int(input("Enter the quantity to be changed : "))
products[update_product]=update_quantity
print(products)
#take a threshold
threshhold=int(input("Enter the threshold : "))
for x,y in products.items():
  if y<threshhold:
    print(x, y)
prod=input("Enter the productto be deleted : ")
del products[prod]
print(products)
```

Classes and objects

Abstract methods

```python
from abc import *
import math
@abstractmethod
class number:
  def calculate(self,x):
    pass
class square(number):
  def calculate(self,x):
    return x*x
  def display(self):
    print("Hello World")
class cube(number):
  def calculate(self,x):
    return x*x*x
class square_root(number):
  def calculate(self,x):
    return math.sqrt(x)
obj1=square()
print(obj1.calculate(int(input("Enter a number to find the square : "))))
obj1.display()
obj2=cube()
print(obj2.calculate(int(input("Enter the number to find the cube : "))))
obj3=square_root()
print(obj3.calculate(int(input("enter the number to find the square root : "))))
```

Create a class called Leader. Derive this class from the
base classes Person and Employee.
The attributes of Person class are name and idnumber.
The attributes of Employee class are salary and post.

```python
class Person:
    def __init__(self, name, idnumber):
        self.name = name
        self.idnumber = idnumber


class Employee:
    def __init__(self, salary, post):
        self.salary = salary
        self.post = post


class Leader(Person, Employee):
    def __init__(self, name, idnumber, salary, post):
        Person.__init__(self, name, idnumber)
        Employee.__init__(self, salary, post)

    def display(self):
        print("Name:", self.name)
        print("ID:", self.idnumber)
        print("Salary:", self.salary)
        print("Post:", self.post)


# Object creation
l = Leader("Krishna", 101, 50000, "Manager")
l.display()
```
